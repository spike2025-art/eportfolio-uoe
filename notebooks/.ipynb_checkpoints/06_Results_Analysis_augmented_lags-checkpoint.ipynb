# 06 — Results Analysis  *(augmented_lags version)*

Loads all outputs from `05_Model_Training_augmented_lags` and produces dissertation-ready figures.

Input:  `data/processed/augmented_lags/`  
Output: `figures/augmented_lags/` (additional dissertation figures)


## Cell 1 — Imports and Load

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from pathlib import Path
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss
import shap
import xgboost as xgb

BASE    = Path(r'C:\Users\Owner\OneDrive\dissertation')
AUG_DIR = BASE / 'data' / 'processed' / 'augmented_lags'
FIG_DIR = BASE / 'figures' / 'augmented_lags'
FIG_DIR.mkdir(parents=True, exist_ok=True)

# Load all augmented_lags outputs
df_model    = pd.read_csv(AUG_DIR / 'df_model_aug.csv')
results_df  = pd.read_csv(AUG_DIR / 'model_results.csv')
shap_df     = pd.read_csv(AUG_DIR / 'shap_values_m3.csv')
oos_df      = pd.read_csv(AUG_DIR / 'oos_predictions.csv')
M1_FEATURES = pd.read_csv(AUG_DIR / 'M1_features.csv', header=None)[0].tolist()
M2_FEATURES = pd.read_csv(AUG_DIR / 'M2_features.csv', header=None)[0].tolist()
M3_FEATURES = pd.read_csv(AUG_DIR / 'M3_features.csv', header=None)[0].tolist()

TARGET = next(c for c in df_model.columns if 'crisis' in c.lower())

print(f'Model dataset: {df_model.shape}')
print(f'SHAP values:   {shap_df.shape}')
print(f'OOS preds:     {oos_df.shape}')
print()
print(results_df.to_string(index=False))
```

## Cell 2 — Metrics Summary with Bootstrap CIs

```python
from sklearn.utils import resample

def bootstrap_auroc(y_true, y_prob, n_boot=1000, seed=42):
    rng = np.random.RandomState(seed)
    boot_scores = []
    for _ in range(n_boot):
        idx = resample(np.arange(len(y_true)), random_state=rng)
        if y_true[idx].sum() == 0:
            continue
        boot_scores.append(roc_auc_score(y_true[idx], y_prob[idx]))
    return np.percentile(boot_scores, [2.5, 97.5])

base_rate = df_model[TARGET].mean()
print('=' * 72)
print(f' {"Model":<38} {"AUROC":>7} {"95% CI":>14} {"AUPRC":>7} {"Brier":>7}')
print('-' * 72)

model_data = {
    'M1_Logistic_macro_lag1':       (oos_df['y_m1'].dropna(), oos_df['prob_m1'].dropna()),
    'M2_RandomForest_macro_lag123': (oos_df['y_m1'].dropna(), oos_df['prob_m2'].dropna()),
    'M3_XGBoost_sent_lag12':        (oos_df['y_m1'].dropna(), oos_df['prob_m3'].dropna()),
}
for name, (yt, yp) in model_data.items():
    yt = yt.values; yp = yp.values
    mask = ~np.isnan(yt) & ~np.isnan(yp)
    yt, yp = yt[mask], yp[mask]
    if yt.sum() == 0: continue
    auroc = roc_auc_score(yt, yp)
    auprc = average_precision_score(yt, yp)
    brier = brier_score_loss(yt, yp)
    ci_lo, ci_hi = bootstrap_auroc(yt, yp)
    print(f' {name:<38} {auroc:>7.4f} [{ci_lo:.3f},{ci_hi:.3f}] {auprc:>7.4f} {brier:>7.4f}')
print('=' * 72)
print(f' Baseline AUPRC (no-skill): {base_rate:.4f}')
```

## Cell 3 — SHAP by Lag Depth Bar Chart

```python
shap_values = shap_df.values
mean_abs    = pd.Series(np.abs(shap_values).mean(axis=0), index=M3_FEATURES)

# Rebuild LAG_GROUPS from feature names
def get_group(feat):
    if feat.endswith('_lag1_dm') and any(s in feat for s in ['P_pos','P_neg','P_neutral','net_sentiment']):
        return 'sent_lag1'
    if feat.endswith('_lag2_dm') and any(s in feat for s in ['P_pos','P_neg','P_neutral','net_sentiment']):
        return 'sent_lag2'
    if feat.endswith('_lag1_dm'): return 'macro_lag1'
    if feat.endswith('_lag2_dm'): return 'macro_lag2'
    if feat.endswith('_lag3_dm'): return 'macro_lag3'
    return 'other'

groups = pd.Series({f: get_group(f) for f in M3_FEATURES})
lag_imp = mean_abs.groupby(groups).sum().sort_values(ascending=False)

palette = {'macro_lag1':'#1F4E79','macro_lag2':'#2E75B6','macro_lag3':'#9DC3E6',
           'sent_lag1':'#C55A11','sent_lag2':'#F4B183','other':'#AAAAAA'}

fig, ax = plt.subplots(figsize=(8, 4))
ax.barh(lag_imp.index[::-1], lag_imp.values[::-1],
        color=[palette.get(g,'#999') for g in lag_imp.index[::-1]])
ax.set_xlabel('Sum of Mean |SHAP| values', fontsize=11)
ax.set_title('M3 Predictive Signal by Lag Horizon\n'
             '(dark blue = macro, orange = FinBERT sentiment)', fontsize=11, fontweight='bold')
ax.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.savefig(FIG_DIR / 'fig_shap_lag_depth.png', dpi=150, bbox_inches='tight')
plt.show()
print(lag_imp.to_string())
```

## Cell 4 — Temporal Crisis Probability Analysis

```python
oos_temp = oos_df.copy()
oos_temp['year'] = oos_temp['year'].astype(int)
yearly = oos_temp.groupby('year').agg(
    prob_m1=('prob_m1','mean'),
    prob_m2=('prob_m2','mean'),
    prob_m3=('prob_m3','mean'),
    actual_crises=(TARGET,'sum')
).reset_index()

fig, ax1 = plt.subplots(figsize=(12, 5))
ax2 = ax1.twinx()

ax1.plot(yearly['year'], yearly['prob_m1'], color='#2196F3', lw=2, label='M1 Logistic', ls='--')
ax1.plot(yearly['year'], yearly['prob_m2'], color='#4CAF50', lw=2, label='M2 Random Forest')
ax1.plot(yearly['year'], yearly['prob_m3'], color='#FF5722', lw=2, label='M3 XGBoost+Sentiment', ls=':')
ax2.bar(yearly['year'], yearly['actual_crises'], alpha=0.25, color='red', label='Actual crises')

ax1.set_xlabel('Year'); ax1.set_ylabel('Mean Predicted Crisis Probability')
ax2.set_ylabel('Crisis Events', color='red')
ax1.set_title('Out-of-Sample Crisis Probability Over Time (augmented_lags)',
              fontsize=12, fontweight='bold')
l1,lb1 = ax1.get_legend_handles_labels()
l2,lb2 = ax2.get_legend_handles_labels()
ax1.legend(l1+l2, lb1+lb2, loc='upper left', fontsize=9)
ax1.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(FIG_DIR / 'fig_temporal_probs.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Cell 5 — Top-20 SHAP Feature Importance Bar Chart

```python
mean_abs_sorted = mean_abs.sort_values(ascending=True).tail(20)

def feat_color(f):
    if any(s in f for s in ['P_pos','P_neg','P_neutral','net_sentiment']): return '#C55A11'
    if '_lag3' in f: return '#9DC3E6'
    if '_lag2' in f: return '#2E75B6'
    return '#1F4E79'

fig, ax = plt.subplots(figsize=(9, 7))
ax.barh(mean_abs_sorted.index, mean_abs_sorted.values,
        color=[feat_color(f) for f in mean_abs_sorted.index])
ax.set_xlabel('Mean |SHAP value|', fontsize=11)
ax.set_title('Top-20 Features by SHAP Importance — M3 (augmented_lags)\n'
             'dark blue=macro lag1, mid blue=lag2, light blue=lag3, orange=sentiment',
             fontsize=10, fontweight='bold')
ax.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.savefig(FIG_DIR / 'fig_shap_top20.png', dpi=150, bbox_inches='tight')
plt.show()
```
