# 08 — Robustness Checks  *(addresses examiner critique)*

**Purpose:** Three targeted robustness analyses to address the critical assessment of Chapter 4:

1. **Strict temporal validation** — pre-GFC train (2003-2006), GFC test (2007-2009). If M2 AUROC drops sharply, current results are inflated by temporal leakage.
2. **M1b: Logistic Regression with lag1-3** — isolates the lag effect from the non-linearity effect, fixing the identification problem in the M1→M2 comparison.
3. **Horizon × sentiment uplift table** — M2 vs M3 AUPRC at H1, H2, H3. Answers whether sentiment adds value at longer horizons even if not at shorter ones.

Input:  `data/processed/sentiment_improved/df_model_imp.csv`
Output: `data/processed/robustness/` | `figures/robustness/`


## Cell 1 — Imports and Load

```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss
import xgboost as xgb

BASE    = Path(r'C:\Users\Owner\OneDrive\dissertation')
IMP_DIR = BASE / 'data' / 'processed' / 'sentiment_improved'
ROB_DIR = BASE / 'data' / 'processed' / 'robustness'
FIG_DIR = BASE / 'figures' / 'robustness'
ROB_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)
RANDOM_STATE = 42

# Load model dataset from sentiment_improved run
df_model    = pd.read_csv(IMP_DIR / 'df_model_imp.csv')
M1_FEATURES = pd.read_csv(IMP_DIR / 'M1_features.csv', header=None)[0].tolist()
M2_FEATURES = pd.read_csv(IMP_DIR / 'M2_features.csv', header=None)[0].tolist()
M3_FEATURES = pd.read_csv(IMP_DIR / 'M3_features.csv', header=None)[0].tolist()

# Primary target
TARGET = next(c for c in df_model.columns if 'target' in c.lower())
if TARGET not in df_model.columns:
    TARGET = next(c for c in df_model.columns if 'crisis' in c.lower()
                  and c not in ['crisisjst','crisisjst_old'])

print(f'Dataset: {df_model.shape}')
print(f'Target : {TARGET}')
print(f'Events : {int(df_model[TARGET].sum())} / {len(df_model)}')
print(f'Years  : {df_model["year"].min()} - {df_model["year"].max()}')
print()
print(f'M1 features: {len(M1_FEATURES)}')
print(f'M2 features: {len(M2_FEATURES)}')
print(f'M3 features: {len(M3_FEATURES)}')
```

## Cell 2 — M1b: Logistic Regression with lag1-3 features

**Why this matters:** The examiner correctly notes that M1→M2 changes two things simultaneously:
the model (logistic → RF) and the features (lag1 only → lag1-3). This makes it impossible to
attribute the AUPRC gain to non-linearity vs additional lags.

M1b uses the same logistic regression as M1 but with the full M2 feature set (lag1-3 macro).
The comparison table then cleanly isolates:
- M1 → M1b: pure lag effect (same model, more lags)
- M1b → M2: pure non-linearity effect (same features, different model)


```python
# M1b uses M2 features (lag1-3 macro) with a logistic regression
# This isolates the lag contribution from the non-linearity contribution

scale_pw = (df_model[TARGET]==0).sum() / (df_model[TARGET]==1).sum()

def expanding_window_cv(df, feature_cols, target_col, model_factory,
                        min_train_years=4, hold_out_years=(2019,2020)):
    cv_years   = sorted(y for y in df['year'].unique() if y not in hold_out_years)
    first_eval = cv_years[min_train_years]
    eval_years = [y for y in cv_years if y >= first_eval]
    oos_idx, oos_prob, records = [], [], []
    for yr in eval_years:
        tr = df['year'] < yr; te = df['year'] == yr
        Xtr,ytr = df.loc[tr,feature_cols], df.loc[tr,target_col]
        Xte,yte = df.loc[te,feature_cols], df.loc[te,target_col]
        if ytr.sum()==0 or len(Xte)==0: continue
        m = model_factory(); m.fit(Xtr,ytr)
        prob = m.predict_proba(Xte)[:,1]
        oos_idx.extend(df.index[te].tolist())
        oos_prob.extend(prob.tolist())
        records.append({'year':yr,'crises_train':int(ytr.sum()),'crises_test':int(yte.sum())})
    y_oos = df.loc[oos_idx,target_col].values
    return y_oos, np.array(oos_prob), oos_idx, pd.DataFrame(records)

# M1: LR lag1 only (original)
def m1_f(): return Pipeline([('sc',StandardScaler()),
    ('clf',LogisticRegression(max_iter=1000,class_weight='balanced',C=0.1,random_state=RANDOM_STATE))])

# M1b: LR with lag1-3 (new — isolates lag effect)
def m1b_f(): return Pipeline([('sc',StandardScaler()),
    ('clf',LogisticRegression(max_iter=1000,class_weight='balanced',C=0.1,random_state=RANDOM_STATE))])

# M2: RF with lag1-3 (original)
def m2_f(): return RandomForestClassifier(n_estimators=500,max_depth=5,
    class_weight='balanced',random_state=RANDOM_STATE,n_jobs=-1)

# M3: XGB with macro+sent
def m3_f(): return xgb.XGBClassifier(n_estimators=300,max_depth=4,learning_rate=0.05,
    scale_pos_weight=scale_pw,random_state=RANDOM_STATE,eval_metric='logloss',verbosity=0)

print('Training M1  (LR, lag1)...')
y_m1, p_m1, _,_ = expanding_window_cv(df_model, M1_FEATURES, TARGET, m1_f)
print('Training M1b (LR, lag1-3)...')
y_m1b,p_m1b,_,_ = expanding_window_cv(df_model, M2_FEATURES, TARGET, m1b_f)
print('Training M2  (RF, lag1-3)...')
y_m2, p_m2, _,_ = expanding_window_cv(df_model, M2_FEATURES, TARGET, m2_f)
print('Training M3  (XGB+sent)...')
y_m3, p_m3, _,_ = expanding_window_cv(df_model, M3_FEATURES, TARGET, m3_f)

base_rate = df_model[TARGET].mean()

# Build the identification table
ident_rows = []
for name, yt, yp in [
        ('M1: LR (lag1 only)',      y_m1,  p_m1),
        ('M1b: LR (lag1-3)',        y_m1b, p_m1b),
        ('M2: RF (lag1-3)',         y_m2,  p_m2),
        ('M3: XGB (macro+sent)',    y_m3,  p_m3),
]:
    auroc = roc_auc_score(yt, yp)
    auprc = average_precision_score(yt, yp)
    brier = brier_score_loss(yt, yp)
    lift  = auprc / base_rate
    ident_rows.append({'Model':name,'AUROC':round(auroc,4),'AUPRC':round(auprc,4),
                       'AUPRC/Base':f'{lift:.1f}x','Brier':round(brier,4)})

ident_df = pd.DataFrame(ident_rows)
print()
print('=== IDENTIFICATION TABLE ===')
print(ident_df.to_string(index=False))
print()
print('M1 -> M1b: pure LAG EFFECT (same LR model, more lags)')
print(f'  AUPRC change: {ident_rows[0]["AUPRC"]} -> {ident_rows[1]["AUPRC"]}',
      f'({ident_rows[1]["AUPRC"]-ident_rows[0]["AUPRC"]:+.4f})')
print()
print('M1b -> M2: pure NON-LINEARITY EFFECT (same lag1-3 features, LR -> RF)')
print(f'  AUPRC change: {ident_rows[1]["AUPRC"]} -> {ident_rows[2]["AUPRC"]}',
      f'({ident_rows[2]["AUPRC"]-ident_rows[1]["AUPRC"]:+.4f})')

ident_df.to_csv(ROB_DIR / 'identification_table.csv', index=False)
print(f'Saved identification_table.csv')
```

## Cell 3 — Plot Identification Analysis


```python
models_id = ['M1\nLR lag1', 'M1b\nLR lag1-3', 'M2\nRF lag1-3', 'M3\nXGB+sent']
aurocs_id  = [ident_rows[i]['AUROC'] for i in range(4)]
auprcs_id  = [ident_rows[i]['AUPRC'] for i in range(4)]
colors_id  = ['#9DC3E6','#2E75B6','#1F4E79','#C55A11']

fig, (ax1,ax2) = plt.subplots(1,2,figsize=(12,5))

bars1 = ax1.bar(models_id, aurocs_id, color=colors_id, width=0.5)
ax1.set_ylim(0.8,1.02)
ax1.set_ylabel('AUROC',fontsize=11)
ax1.set_title('AUROC: Lag Effect vs Non-Linearity Effect',fontweight='bold')
ax1.grid(True,alpha=0.3,axis='y')
for bar,v in zip(bars1,aurocs_id): ax1.text(bar.get_x()+bar.get_width()/2,v+0.003,f'{v:.3f}',ha='center',fontsize=9)
ax1.axvline(1.5,color='red',linestyle='--',alpha=0.5,linewidth=1)
ax1.text(0.5,0.88,'Lag effect',ha='center',transform=ax1.get_xaxis_transform(),color='#2E75B6',fontsize=9)
ax1.text(2.0,0.88,'Non-linearity',ha='center',transform=ax1.get_xaxis_transform(),color='#1F4E79',fontsize=9)

bars2 = ax2.bar(models_id, auprcs_id, color=colors_id, width=0.5)
ax2.axhline(base_rate,color='k',linestyle=':',alpha=0.5,label=f'No-skill ({base_rate:.3f})')
ax2.set_ylabel('AUPRC',fontsize=11)
ax2.set_title('AUPRC: Lag Effect vs Non-Linearity Effect',fontweight='bold')
ax2.grid(True,alpha=0.3,axis='y')
ax2.legend(fontsize=9)
for bar,v in zip(bars2,auprcs_id): ax2.text(bar.get_x()+bar.get_width()/2,v+0.01,f'{v:.3f}',ha='center',fontsize=9)
ax2.axvline(1.5,color='red',linestyle='--',alpha=0.5,linewidth=1)

plt.suptitle('Identification: Decomposing the M1 → M2 Performance Gain',
             fontsize=12,fontweight='bold')
plt.tight_layout()
plt.savefig(FIG_DIR/'fig_identification.png',dpi=150,bbox_inches='tight')
plt.show()
print(f'Saved fig_identification.png')
```

## Cell 4 — Strict Temporal Validation: Pre-GFC Train → GFC Test

**Why this matters:** The examiner questions whether AUROC 0.984 is inflated.
The expanding-window CV has one fold with crisis cases (year 2007).
A strict out-of-sample test trains ONLY on pre-crisis data and tests ONLY on the crisis period.

Two splits:
- **Split A**: Train 2003-2005 → Test 2006-2009 (pre-GFC build-up train, GFC test)
- **Split B**: Train 2003-2007 → Test 2008-2012 (includes GFC onset in train, post-GFC test)

If results hold under Split A, the model is genuinely predictive.
If they collapse, the results are driven by GFC-period features in training.


```python

TARGET_STRICT = 'target_h2'

# Override TARGET for strict validation — force target_h2
TARGET = TARGET_STRICT   # target_h2 has events in 2005 (2yr before 2007 onset)


def strict_split_eval(df, feature_cols, target_col, model_factory,
                      train_years, test_years, label):
    tr = df['year'].isin(train_years)
    te = df['year'].isin(test_years)
    Xtr,ytr = df.loc[tr,feature_cols], df.loc[tr,target_col]
    Xte,yte = df.loc[te,feature_cols], df.loc[te,target_col]
    print(f'{label}:')
    print(f'  Train: {train_years[0]}-{train_years[-1]}  n={len(Xtr)}  crises={int(ytr.sum())}')
    print(f'  Test : {test_years[0]}-{test_years[-1]}   n={len(Xte)}  crises={int(yte.sum())}')
    if ytr.sum()==0 or yte.sum()==0:
        print('  SKIP: no positive cases in train or test')
        return None
    m = model_factory(); m.fit(Xtr,ytr)
    prob = m.predict_proba(Xte)[:,1]
    auroc = roc_auc_score(yte,prob)
    auprc = average_precision_score(yte,prob)
    brier = brier_score_loss(yte,prob)
    print(f'  AUROC={auroc:.4f}  AUPRC={auprc:.4f}  Brier={brier:.4f}')
    print()
    return {'label':label,'auroc':auroc,'auprc':auprc,'brier':brier,
            'n_train':len(Xtr),'n_test':len(Xte),'crises_train':int(ytr.sum()),'crises_test':int(yte.sum())}

train_A = list(range(2003,2006))  # 2003-2005: pure pre-crisis
test_A  = list(range(2006,2010))  # 2006-2009: GFC build-up + onset
train_B = list(range(2003,2008))  # 2003-2007: includes crisis build-up
test_B  = list(range(2008,2013))  # 2008-2012: GFC onset + aftermath

strict_results = []
print('=== STRICT TEMPORAL VALIDATION ===')
print()

for split_label, tr_yrs, te_yrs in [
    ('Split A: Pre-crisis train | GFC test', train_A, test_A),
    ('Split B: GFC train | Post-GFC test',   train_B, test_B),
]:
    print(f'--- {split_label} ---')
    for model_label, feat_cols, factory in [
        ('M2_RF',      M2_FEATURES, m2_f),
        ('M3_XGB_sent',M3_FEATURES, m3_f),
    ]:
        res = strict_split_eval(df_model, feat_cols, TARGET, factory,
                                tr_yrs, te_yrs, f'  {model_label}')
        if res:
            res['split'] = split_label
            res['model'] = model_label
            strict_results.append(res)

strict_df = pd.DataFrame(strict_results)
strict_df.to_csv(ROB_DIR/'strict_temporal_validation.csv', index=False)
print('Saved strict_temporal_validation.csv')
print()
print('INTERPRETATION GUIDE:')
print('If AUROC in Split A is >= 0.75: results are genuinely predictive')
print('If AUROC in Split A is < 0.65: expanding-window results may be inflated')
```

## Cell 5 — Strict Validation Chart

```python
if len(strict_df) >= 2:
    fig, (ax1,ax2) = plt.subplots(1,2,figsize=(12,5))

    for ax, metric, ylabel in [(ax1,'auroc','AUROC'),(ax2,'auprc','AUPRC')]:
        for model, color, marker in [('M2_RF','#1F4E79','o'),('M3_XGB_sent','#C55A11','s')]:
            sub = strict_df[strict_df['model']==model]
            vals = sub[metric].values
            labels = ['Split A\n(pre-GFC train)','Split B\n(GFC train)'][:len(vals)]
            x = np.arange(len(vals))
            ax.plot(x, vals, marker=marker, color=color, lw=2,
                    markersize=10, label=model)
            for xi,v in zip(x,vals): ax.annotate(f'{v:.3f}',(xi,v),
                textcoords='offset points',xytext=(0,10),ha='center',fontsize=9)
        ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels)
        ax.set_ylabel(ylabel,fontsize=11)
        ax.set_title(f'Strict Temporal Validation — {ylabel}',fontweight='bold')
        ax.legend(fontsize=9); ax.grid(True,alpha=0.3)
        if metric=='auroc': ax.axhline(0.75,color='grey',linestyle=':',
            alpha=0.5,label='Literature benchmark (0.75)')
        if metric=='auprc': ax.axhline(base_rate,color='k',linestyle=':',
            alpha=0.4,label=f'No-skill ({base_rate:.3f})')
        ax.legend(fontsize=8)

    plt.suptitle('Strict Temporal Validation: Pre-GFC Train → GFC Test',
                 fontsize=12,fontweight='bold')
    plt.tight_layout()
    plt.savefig(FIG_DIR/'fig_strict_temporal.png',dpi=150,bbox_inches='tight')
    plt.show()
    print(f'Saved fig_strict_temporal.png')
else:
    print('Insufficient results to plot — check crisis counts above')
```

## Cell 6 — Horizon × Sentiment Uplift Table

**Why this matters:** The examiner identifies this as the key missed contribution.
Does FinBERT sentiment add value at longer horizons even if not at shorter ones?

Runs M2 (macro only) and M3 (macro+sent) separately at each of H1, H2, H3.
The AUPRC uplift (M3 - M2) by horizon is the core question.


```python
# Get all horizon target columns
horizon_targets = {
    'H1_1yr': 'target_h1',
    'H2_2yr': 'target_h2',
    'H3_3yr': 'target_h3',
}

# Check which are present
available = {k:v for k,v in horizon_targets.items() if v in df_model.columns}
print(f'Available horizon targets: {list(available.keys())}')
print()

uplift_rows = []

for h_name, h_target in available.items():
    spw_h = (df_model[h_target]==0).sum()/(df_model[h_target]==1).sum()

    def m2_h_f(): return RandomForestClassifier(
        n_estimators=500,max_depth=5,class_weight='balanced',
        random_state=RANDOM_STATE,n_jobs=-1)

    def m3_h_f(spw=spw_h): return xgb.XGBClassifier(
        n_estimators=300,max_depth=4,learning_rate=0.05,scale_pos_weight=spw,
        random_state=RANDOM_STATE,eval_metric='logloss',verbosity=0)

    print(f'Training M2 and M3 for {h_name}...')
    y_m2h,p_m2h,_,_ = expanding_window_cv(df_model,M2_FEATURES,h_target,m2_h_f)
    y_m3h,p_m3h,_,_ = expanding_window_cv(df_model,M3_FEATURES,h_target,m3_h_f)

    if y_m2h.sum()==0 or y_m3h.sum()==0:
        print(f'  Skipping {h_name}: no positive cases in OOS')
        continue

    base_h  = y_m2h.mean()
    auprc_m2 = average_precision_score(y_m2h,p_m2h)
    auprc_m3 = average_precision_score(y_m3h,p_m3h)
    auroc_m2 = roc_auc_score(y_m2h,p_m2h)
    auroc_m3 = roc_auc_score(y_m3h,p_m3h)
    uplift   = auprc_m3 - auprc_m2
    pct_upl  = 100*uplift/auprc_m2 if auprc_m2>0 else 0

    uplift_rows.append({
        'Horizon':    h_name,
        'M2_AUROC':   round(auroc_m2,4),
        'M3_AUROC':   round(auroc_m3,4),
        'M2_AUPRC':   round(auprc_m2,4),
        'M3_AUPRC':   round(auprc_m3,4),
        'Uplift':     round(uplift,4),
        'Uplift_pct': round(pct_upl,1),
        'Base_rate':  round(base_h,4),
        'N_events':   int(y_m2h.sum())
    })
    print(f'  M2 AUPRC={auprc_m2:.4f}  M3 AUPRC={auprc_m3:.4f}  Uplift={uplift:+.4f} ({pct_upl:+.1f}%)')

uplift_df = pd.DataFrame(uplift_rows)
print()
print('=== HORIZON x SENTIMENT UPLIFT TABLE ===')
print(uplift_df[['Horizon','M2_AUPRC','M3_AUPRC','Uplift','Uplift_pct','Base_rate','N_events']].to_string(index=False))
print()
print('Positive uplift = sentiment helps at that horizon')
print('Negative uplift = sentiment hurts at that horizon')
uplift_df.to_csv(ROB_DIR/'horizon_sentiment_uplift.csv', index=False)
print(f'Saved horizon_sentiment_uplift.csv')
```

## Cell 7 — Horizon Uplift Chart

```python
if len(uplift_df) >= 2:
    horizons  = uplift_df['Horizon'].tolist()
    m2_auprcs = uplift_df['M2_AUPRC'].tolist()
    m3_auprcs = uplift_df['M3_AUPRC'].tolist()
    uplifts   = uplift_df['Uplift'].tolist()
    x = np.arange(len(horizons)); w=0.3

    fig,(ax1,ax2) = plt.subplots(1,2,figsize=(13,5))

    # Side-by-side M2 vs M3 AUPRC by horizon
    b1 = ax1.bar(x-w/2, m2_auprcs, w, color='#1F4E79', label='M2 (macro only)')
    b2 = ax1.bar(x+w/2, m3_auprcs, w, color='#C55A11', label='M3 (macro+sentiment)')
    ax1.axhline(float(uplift_df['Base_rate'].mean()),color='k',linestyle=':',alpha=0.4,label='No-skill')
    ax1.set_xticks(x); ax1.set_xticklabels(horizons)
    ax1.set_ylabel('AUPRC',fontsize=11)
    ax1.set_title('M2 vs M3 AUPRC by Prediction Horizon',fontweight='bold')
    ax1.legend(fontsize=9); ax1.grid(True,alpha=0.3,axis='y')
    for b,v in zip(list(b1)+list(b2), m2_auprcs+m3_auprcs):
        ax1.text(b.get_x()+b.get_width()/2,v+0.01,f'{v:.3f}',ha='center',fontsize=8)

    # Uplift by horizon
    colors_up = ['#2E75B6' if u>=0 else '#d32f2f' for u in uplifts]
    ax2.bar(horizons, uplifts, color=colors_up, width=0.5)
    ax2.axhline(0,color='k',linewidth=1)
    ax2.set_ylabel('AUPRC Uplift (M3 − M2)',fontsize=11)
    ax2.set_title('Sentiment AUPRC Uplift by Horizon\n(blue=helps, red=hurts)',fontweight='bold')
    ax2.grid(True,alpha=0.3,axis='y')
    for i,(h,u) in enumerate(zip(horizons,uplifts)):
        ax2.text(i,u+(0.005 if u>=0 else -0.015),f'{u:+.3f}',ha='center',fontsize=10,fontweight='bold')

    plt.suptitle('Horizon x Sentiment Uplift Analysis',fontsize=12,fontweight='bold')
    plt.tight_layout()
    plt.savefig(FIG_DIR/'fig_horizon_sentiment_uplift.png',dpi=150,bbox_inches='tight')
    plt.show()
    print(f'Saved fig_horizon_sentiment_uplift.png')
```
