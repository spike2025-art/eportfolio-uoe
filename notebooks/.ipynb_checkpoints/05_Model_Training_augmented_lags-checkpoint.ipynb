# 05 — Model Training  *(augmented_lags version)*

**What is new vs the original notebook 05:**
- Macro features lagged at **t−1, t−2, t−3** (not current-year unlagged values)
- Sentiment features at **t−1 and t−2** (from updated notebook 04)
- All features country-demeaned **after** lagging
- Three clean model specs: M1 (LR, macro lag1), M2 (RF, macro lag1-3), M3 (XGB, macro+sent lags)
- **Expanding-window** cross-validation (time-safe — no future data in training)
- Per-fold SHAP collection for stability analysis
- All outputs saved to `data/processed/augmented_lags/` and `figures/augmented_lags/`

Input:  `data/processed/jst_sentiment_master.csv`  
Output: `data/processed/augmented_lags/model_results.csv`, `shap_values_m3.csv`, `df_model_aug.csv`


## Cell 1 — Imports and Paths

```python
import os
os.environ['PYTHONWARNINGS'] = 'ignore'

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss
import xgboost as xgb
import shap
import tempfile, os as _os

# ── PATHS ────────────────────────────────────────────────────
BASE         = Path(r'C:\Users\Owner\OneDrive\dissertation')
DATA_IN      = BASE / 'data' / 'processed' / 'jst_sentiment_master.csv'
OUT_DIR      = BASE / 'data' / 'processed' / 'augmented_lags'
FIG_DIR      = BASE / 'figures' / 'augmented_lags'
OUT_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)
RANDOM_STATE = 42

assert DATA_IN.exists(), f'Input not found: {DATA_IN}'
print('=' * 65)
print(' 05 — MODEL TRAINING (augmented_lags)')
print('=' * 65)
print(f'Input  : {DATA_IN}')
print(f'OutDir : {OUT_DIR}')
print(f'FigDir : {FIG_DIR}')
print()
print(f'numpy   {np.__version__}  |  xgboost {xgb.__version__}  |  shap {shap.__version__}')
```

## Cell 2 — Load Master Dataset

```python
df_raw = pd.read_csv(DATA_IN)
# CRITICAL: sort before any shift() — wrong sort = wrong lags
df_raw = df_raw.sort_values(['iso', 'year']).reset_index(drop=True)
df_raw.columns = df_raw.columns.str.strip()

print(f'Shape      : {df_raw.shape}')
print(f'Countries  : {df_raw["iso"].nunique()} — {sorted(df_raw["iso"].unique())}')
print(f'Year range : {df_raw["year"].min()} – {df_raw["year"].max()}')
TARGET = next(c for c in df_raw.columns if 'crisis' in c.lower())
print(f'Target col : {TARGET}  |  events: {int(df_raw[TARGET].sum())} / {len(df_raw)}')
print()
print('All columns:')
print(df_raw.columns.tolist())
```

## Cell 3 — Define Macro Feature Groups

```python
# JST macro level variables — script checks which are actually present
MACRO_LEVELS_CANDIDATES = [
    'tloans',    # total loans/credit
    'tmort',     # mortgage loans
    'money',     # broad money
    'ca',        # current account/GDP
    'ltrate',    # long-term interest rate
    'stir',      # short-term interest rate
    'hpnom',     # nominal house prices
    'debtgdp',   # government debt/GDP
    'cpi',       # consumer price index
    'rgdpmad',   # real GDP per capita
]

MACRO_LEVELS = [f for f in MACRO_LEVELS_CANDIDATES if f in df_raw.columns]

# Sentiment lag columns already in master CSV (from modified notebook 04)
SENT_LAG1 = [c for c in df_raw.columns if c.endswith('_lag1') and
             any(s in c for s in ['P_pos','P_neg','P_neutral','net_sentiment'])]
SENT_LAG2 = [c for c in df_raw.columns if c.endswith('_lag2') and
             any(s in c for s in ['P_pos','P_neg','P_neutral','net_sentiment'])]

print(f'Macro level cols ({len(MACRO_LEVELS)}): {MACRO_LEVELS}')
print(f'Sentiment lag1 cols ({len(SENT_LAG1)}): {SENT_LAG1}')
print(f'Sentiment lag2 cols ({len(SENT_LAG2)}): {SENT_LAG2}')

if not SENT_LAG2:
    print()
    print('WARNING: lag2 sentiment not found in master CSV.')
    print('Run the modified 04_Sentiment_Aggregation_augmented_lags first.')
```

## Cell 4 — Compute Growth Rates (within-country, year t)

```python
df = df_raw.copy()

# Growth-rate features — still year-t values; will be lagged in Cell 5
GROWTH_FEATURES = []
for col in ['tloans', 'tmort', 'money', 'hpnom', 'rgdpmad']:
    if col in df.columns:
        gr_col = f'{col}_gr'
        df[gr_col] = df.groupby('iso')[col].pct_change()
        GROWTH_FEATURES.append(gr_col)

# Term spread
if 'ltrate' in df.columns and 'stir' in df.columns:
    df['term_spread'] = df['ltrate'] - df['stir']
    GROWTH_FEATURES.append('term_spread')

# All macro features to be lagged
MACRO_ALL = MACRO_LEVELS + GROWTH_FEATURES
MACRO_ALL = [f for f in MACRO_ALL if f in df.columns]

print(f'Growth features computed: {GROWTH_FEATURES}')
print(f'All macro features to lag ({len(MACRO_ALL)}): {MACRO_ALL}')
```

## Cell 5 — Generate Multi-Lag Features (the core augmentation)

```python
# ── Multi-lag generation ─────────────────────────────────────────────────────
# Macro: t-1, t-2, t-3    |    Sentiment: already lagged in master CSV (t-1, t-2)
# All shifts are within-country — cross-country contamination is impossible

def add_lags(df, features, lags, group_col='iso'):
    df = df.copy()
    for feat in features:
        if feat not in df.columns:
            continue
        for lag in lags:
            df[f'{feat}_lag{lag}'] = df.groupby(group_col)[feat].shift(lag)
    return df

df = add_lags(df, MACRO_ALL, lags=[1, 2, 3])

macro_lag_cols = [f'{feat}_lag{k}' for feat in MACRO_ALL for k in [1,2,3]
                  if f'{feat}_lag{k}' in df.columns]

print(f'Macro lag columns created: {len(macro_lag_cols)}')
print('Sample:', macro_lag_cols[:6], '...')
print()
print(f'Sentiment lag1 cols: {SENT_LAG1}')
print(f'Sentiment lag2 cols: {SENT_LAG2}')
```

## Cell 6 — Country Demean All Lagged Features

```python
# Demean macro lags within country (removes country fixed effects)
ALL_LAG_COLS = macro_lag_cols + SENT_LAG1 + SENT_LAG2

DM_COLS = []
for col in ALL_LAG_COLS:
    dm_col = f'{col}_dm'
    df[dm_col] = df[col] - df.groupby('iso')[col].transform('mean')
    DM_COLS.append(dm_col)

print(f'Demeaned columns created: {len(DM_COLS)}')

# Build feature-group name lists for SHAP decomposition later
MACRO_DM_LAG1 = [f'{f}_lag1_dm' for f in MACRO_ALL if f'{f}_lag1_dm' in df.columns]
MACRO_DM_LAG2 = [f'{f}_lag2_dm' for f in MACRO_ALL if f'{f}_lag2_dm' in df.columns]
MACRO_DM_LAG3 = [f'{f}_lag3_dm' for f in MACRO_ALL if f'{f}_lag3_dm' in df.columns]
SENT_DM_LAG1  = [f'{c}_dm' for c in SENT_LAG1 if f'{c}_dm' in df.columns]
SENT_DM_LAG2  = [f'{c}_dm' for c in SENT_LAG2 if f'{c}_dm' in df.columns]

LAG_GROUPS = {
    'macro_lag1': MACRO_DM_LAG1,
    'macro_lag2': MACRO_DM_LAG2,
    'macro_lag3': MACRO_DM_LAG3,
    'sent_lag1':  SENT_DM_LAG1,
    'sent_lag2':  SENT_DM_LAG2,
}
print()
for grp, cols in LAG_GROUPS.items():
    print(f'  {grp}: {len(cols)} features')
```

## Cell 7 — Drop NaN Rows and Define Feature Sets

```python
# The t-3 lag removes 1997-1999 per country (~54 rows)
all_dm_lag_cols = MACRO_DM_LAG1 + MACRO_DM_LAG2 + MACRO_DM_LAG3 + SENT_DM_LAG1 + SENT_DM_LAG2

df_model = df.dropna(subset=all_dm_lag_cols).copy().reset_index(drop=True)

print(f'Rows: {len(df_raw)} → {len(df_model)} after NaN drop (lost {len(df_raw)-len(df_model)} rows)')
print(f'Year range in model dataset: {df_model["year"].min()} – {df_model["year"].max()}')
print(f'Crisis events: {int(df_model[TARGET].sum())} / {len(df_model)} ({100*df_model[TARGET].mean():.1f}%)')
print()

# ── Three model feature sets (matching dissertation M1/M2/M3) ──
# M1: Logistic Regression — macro lag1 only (replicates Schularick-Taylor 2012)
M1_FEATURES = MACRO_DM_LAG1

# M2: Random Forest — macro all lags (tests non-linearity + multi-horizon)
M2_FEATURES = MACRO_DM_LAG1 + MACRO_DM_LAG2 + MACRO_DM_LAG3

# M3: XGBoost — macro all lags + sentiment both lags (adds FinBERT)
M3_FEATURES = MACRO_DM_LAG1 + MACRO_DM_LAG2 + MACRO_DM_LAG3 + SENT_DM_LAG1 + SENT_DM_LAG2

print(f'M1 features (LR  — macro lag1):          {len(M1_FEATURES)}')
print(f'M2 features (RF  — macro lag1-3):        {len(M2_FEATURES)}')
print(f'M3 features (XGB — macro+sent lag1-2):   {len(M3_FEATURES)}')
print()

# Confirm no current-year features in any set (all must end in _lagN_dm)
for spec, feats in [('M1',M1_FEATURES),('M2',M2_FEATURES),('M3',M3_FEATURES)]:
    bad = [f for f in feats if '_lag' not in f]
    if bad:
        print(f'WARNING {spec}: non-lagged features detected: {bad}')
    else:
        print(f'{spec}: ✓ all features are lagged')
```

## Cell 8 — No-Leakage Verification

```python
# Verify that lag1 in year t matches the raw value in year t-1
print('Running leakage verification...')
errors = 0
sample_feat = MACRO_LEVELS[0]  # e.g. 'tloans'
lag1_col    = f'{sample_feat}_lag1'

if lag1_col in df_model.columns:
    for ctry in df_model['iso'].unique()[:5]:
        sub = df_model[df_model['iso']==ctry][['year', sample_feat, lag1_col]].copy()
        sub = sub.sort_values('year').reset_index(drop=True)
        # lag1 in row i should equal raw value in row i-1
        expected = sub[sample_feat].shift(1)
        mismatch = (sub[lag1_col] - expected).abs().dropna().max()
        status = '\u2705' if mismatch < 1e-6 else '\u274c LEAKAGE'
        print(f'  {ctry}: max lag1 mismatch = {mismatch:.8f}  {status}')
        if mismatch >= 1e-6:
            errors += 1
else:
    print(f'  Skipping: {lag1_col} not in df_model (growth-rate features used)')

print()
print(f'Leakage check complete. Errors: {errors}')
print('All OK — safe to proceed.' if errors == 0 else 'STOP — fix leakage before modelling.')
```

## Cell 9 — Expanding-Window Cross-Validation Function

```python
def expanding_window_cv(df, feature_cols, target_col, model_factory,
                        min_train_years=7, hold_out_years=[2019, 2020]):
    """
    Expanding-window CV for panel data.
    - Trains on all country-years up to year t-1
    - Predicts on all country-years in year t
    - Respects temporal ordering; no future data in training
    - Returns (y_true_oos, y_prob_oos, fold_records)
    """
    years = sorted(df['year'].unique())
    # Remove hold-out years from CV
    cv_years = [y for y in years if y not in hold_out_years]
    train_start = cv_years[0]

    # First eval year = after min_train_years of data
    first_eval = cv_years[min_train_years]

    eval_years = [y for y in cv_years if y >= first_eval]
    print(f'Training burn-in : {train_start} – {first_eval - 1} ({min_train_years} years)')
    print(f'Evaluation years : {eval_years[0]} – {eval_years[-1]} ({len(eval_years)} folds)')
    print(f'Hold-out         : {hold_out_years}')
    print()

    oos_idx, oos_prob = [], []
    fold_records = []

    for eval_yr in eval_years:
        train_mask = df['year'] < eval_yr
        test_mask  = df['year'] == eval_yr

        X_train = df.loc[train_mask, feature_cols]
        y_train = df.loc[train_mask, target_col]
        X_test  = df.loc[test_mask,  feature_cols]
        y_test  = df.loc[test_mask,  target_col]

        if y_train.sum() == 0 or len(X_test) == 0:
            continue

        model = model_factory()
        model.fit(X_train, y_train)

        prob = model.predict_proba(X_test)[:,1]
        oos_idx.extend(df.index[test_mask].tolist())
        oos_prob.extend(prob.tolist())
        fold_records.append({'year': eval_yr, 'n_test': len(y_test),
                             'crises_test': int(y_test.sum()),
                             'n_train': int(train_mask.sum()),
                             'crises_train': int(y_train.sum())})

    y_oos   = df.loc[oos_idx, target_col].values
    df_folds = pd.DataFrame(fold_records)
    return y_oos, np.array(oos_prob), oos_idx, df_folds

print('Expanding-window CV function defined.')
print('Min training: 7 years | Hold-out: 2019-2020')
```

## Cell 10 — Train M1: Logistic Regression (macro lag1)

```python
print('Training M1 — Logistic Regression (macro lag1 only)...')
print(f'Features: {len(M1_FEATURES)}')
print()

def m1_factory():
    return Pipeline([
        ('scaler', StandardScaler()),
        ('clf', LogisticRegression(max_iter=1000, class_weight='balanced',
                                   C=0.1, random_state=RANDOM_STATE))
    ])

y_m1, prob_m1, idx_m1, folds_m1 = expanding_window_cv(
    df_model, M1_FEATURES, TARGET, m1_factory)

auroc_m1 = roc_auc_score(y_m1, prob_m1)
auprc_m1 = average_precision_score(y_m1, prob_m1)
brier_m1 = brier_score_loss(y_m1, prob_m1)

print(f'M1 AUROC  : {auroc_m1:.4f}')
print(f'M1 AUPRC  : {auprc_m1:.4f}  (base rate: {y_m1.mean():.4f})')
print(f'M1 Brier  : {brier_m1:.4f}')
print()
print('Fold summary:')
print(folds_m1.to_string(index=False))
```

## Cell 11 — Train M2: Random Forest (macro lag1-3)

```python
print('Training M2 — Random Forest (macro lag1-3)...')
print(f'Features: {len(M2_FEATURES)}')
print()

def m2_factory():
    return RandomForestClassifier(
        n_estimators=500, max_depth=5,
        class_weight='balanced', random_state=RANDOM_STATE, n_jobs=-1)

y_m2, prob_m2, idx_m2, folds_m2 = expanding_window_cv(
    df_model, M2_FEATURES, TARGET, m2_factory)

auroc_m2 = roc_auc_score(y_m2, prob_m2)
auprc_m2 = average_precision_score(y_m2, prob_m2)
brier_m2 = brier_score_loss(y_m2, prob_m2)

print(f'M2 AUROC  : {auroc_m2:.4f}')
print(f'M2 AUPRC  : {auprc_m2:.4f}')
print(f'M2 Brier  : {brier_m2:.4f}')
```

## Cell 12 — Train M3: XGBoost with FinBERT Sentiment (macro+sent lags)

```python
print('Training M3 — XGBoost (macro lag1-3 + FinBERT lag1-2)...')
print(f'Features: {len(M3_FEATURES)}')
print()

scale_pw = (df_model[TARGET]==0).sum() / (df_model[TARGET]==1).sum()

def m3_factory():
    return xgb.XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        scale_pos_weight=scale_pw, random_state=RANDOM_STATE,
        eval_metric='logloss', verbosity=0)

y_m3, prob_m3, idx_m3, folds_m3 = expanding_window_cv(
    df_model, M3_FEATURES, TARGET, m3_factory)

auroc_m3 = roc_auc_score(y_m3, prob_m3)
auprc_m3 = average_precision_score(y_m3, prob_m3)
brier_m3 = brier_score_loss(y_m3, prob_m3)

print(f'M3 AUROC  : {auroc_m3:.4f}')
print(f'M3 AUPRC  : {auprc_m3:.4f}')
print(f'M3 Brier  : {brier_m3:.4f}')
```

## Cell 13 — Results Comparison Table

```python
base_rate = df_model[TARGET].mean()

results = {
    'M1_Logistic_macro_lag1':      {'AUROC': auroc_m1, 'AUPRC': auprc_m1, 'Brier': brier_m1},
    'M2_RandomForest_macro_lag123':{'AUROC': auroc_m2, 'AUPRC': auprc_m2, 'Brier': brier_m2},
    'M3_XGBoost_sent_lag12':       {'AUROC': auroc_m3, 'AUPRC': auprc_m3, 'Brier': brier_m3},
}

print('=' * 72)
print(f' {"Model":<35} {"AUROC":>7} {"AUPRC":>7} {"AUPRC/Base":>10} {"Brier":>7}')
print('-' * 72)
for name, res in results.items():
    lift = res['AUPRC'] / base_rate
    print(f' {name:<35} {res["AUROC"]:>7.4f} {res["AUPRC"]:>7.4f} {lift:>9.1f}x {res["Brier"]:>7.4f}')
print('=' * 72)
print(f' Baseline AUPRC (no-skill): {base_rate:.4f}  |  Scale_pos_weight: {scale_pw:.1f}')

# Save
results_df = pd.DataFrame(results).T.reset_index().rename(columns={'index':'Model'})
results_df.to_csv(OUT_DIR / 'model_results.csv', index=False)
print(f'\nSaved → {OUT_DIR}/model_results.csv')
```

## Cell 14 — ROC and PR Curves

```python
from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

colors = ['#2196F3', '#4CAF50', '#FF5722']
for (name, res), color, ys, probs in zip(
        results.items(), colors,
        [y_m1, y_m2, y_m3], [prob_m1, prob_m2, prob_m3]):
    RocCurveDisplay.from_predictions(ys, probs, name=name, ax=ax1, color=color)
    PrecisionRecallDisplay.from_predictions(ys, probs, name=name, ax=ax2, color=color)

ax1.plot([0,1],[0,1],'k--',alpha=0.4,label='Random')
ax1.set_title('ROC Curves — augmented_lags', fontweight='bold')
ax1.legend(loc='lower right', fontsize=8)
ax1.grid(True, alpha=0.3)

ax2.axhline(base_rate, color='k', linestyle='--', alpha=0.4, label=f'No-skill ({base_rate:.3f})')
ax2.set_title('Precision-Recall Curves — augmented_lags', fontweight='bold')
ax2.legend(loc='upper right', fontsize=8)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(FIG_DIR / 'roc_pr_curves.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved → {FIG_DIR}/roc_pr_curves.png')
```

## Cell 15 — SHAP on M3 (full-data fit for interpretability)

```python
import tempfile

X_m3 = df_model[M3_FEATURES]
y_all = df_model[TARGET]

xgb_full = xgb.XGBClassifier(
    n_estimators=300, max_depth=4, learning_rate=0.05,
    scale_pos_weight=scale_pw, random_state=RANDOM_STATE,
    eval_metric='logloss', verbosity=0)
xgb_full.fit(X_m3, y_all)

# Base_score fix
tmp = tempfile.mktemp(suffix='.json')
xgb_full.save_model(tmp)
xgb_full.load_model(tmp)
_os.remove(tmp)

explainer   = shap.TreeExplainer(xgb_full)
shap_values = explainer.shap_values(X_m3)
print(f'SHAP values shape: {shap_values.shape}')

# Save
shap_df = pd.DataFrame(shap_values, columns=M3_FEATURES)
shap_df.to_csv(OUT_DIR / 'shap_values_m3.csv', index=False)
print(f'Saved → {OUT_DIR}/shap_values_m3.csv')
```

## Cell 16 — SHAP Beeswarm Summary Plot

```python
plt.figure(figsize=(10, 7))
shap.summary_plot(shap_values, X_m3, feature_names=M3_FEATURES, show=False, plot_type='dot')
plt.title('SHAP Summary — M3 XGBoost (augmented_lags)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(FIG_DIR / 'shap_summary_m3.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved → {FIG_DIR}/shap_summary_m3.png')
```

## Cell 17 — SHAP Importance by Lag Depth (new)

```python
mean_abs_shap = pd.Series(np.abs(shap_values).mean(axis=0), index=M3_FEATURES)

# Aggregate total importance by lag group
lag_importance = {}
for grp, cols in LAG_GROUPS.items():
    valid = [c for c in cols if c in mean_abs_shap.index]
    lag_importance[grp] = mean_abs_shap[valid].sum() if valid else 0.0

lag_series = pd.Series(lag_importance).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 4))
palette = {'macro_lag1':'#1F4E79','macro_lag2':'#2E75B6','macro_lag3':'#9DC3E6',
           'sent_lag1':'#C55A11','sent_lag2':'#F4B183'}
bar_colors = [palette.get(g,'#999999') for g in lag_series.index]
ax.barh(lag_series.index[::-1], lag_series.values[::-1], color=bar_colors[::-1])
ax.set_xlabel('Sum of Mean |SHAP| values', fontsize=11)
ax.set_title('M3 Feature Importance by Lag Depth\n'
             '(blue = macro, orange = FinBERT sentiment)', fontsize=11, fontweight='bold')
ax.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.savefig(FIG_DIR / 'shap_by_lag_depth.png', dpi=150, bbox_inches='tight')
plt.show()

print('Lag depth importance:')
print(lag_series.to_string())
print(f'\nSaved → {FIG_DIR}/shap_by_lag_depth.png')
```

## Cell 18 — Per-Fold SHAP Stability

```python
# Collect SHAP importance per expanding-window fold for M3
print('Computing per-fold SHAP stability (this may take 3-5 minutes)...')

fold_shap_list = []
years_cv = sorted([y for y in df_model['year'].unique() if y not in [2019,2020]])
first_eval = years_cv[7]  # 7-year burn-in

for eval_yr in [y for y in years_cv if y >= first_eval]:
    train_mask = df_model['year'] < eval_yr
    test_mask  = df_model['year'] == eval_yr

    X_tr = df_model.loc[train_mask, M3_FEATURES]
    y_tr = df_model.loc[train_mask, TARGET]
    X_te = df_model.loc[test_mask,  M3_FEATURES]

    if y_tr.sum() == 0 or len(X_te) == 0:
        continue

    m = xgb.XGBClassifier(n_estimators=300, max_depth=4, learning_rate=0.05,
                           scale_pos_weight=scale_pw, random_state=RANDOM_STATE,
                           eval_metric='logloss', verbosity=0)
    m.fit(X_tr, y_tr)

    tmp2 = tempfile.mktemp(suffix='.json')
    m.save_model(tmp2); m.load_model(tmp2); _os.remove(tmp2)

    exp_fold = shap.TreeExplainer(m)
    sv_fold  = exp_fold.shap_values(X_te)
    fold_shap_list.append(pd.Series(np.abs(sv_fold).mean(axis=0), index=M3_FEATURES))
    print(f'  Fold {eval_yr}: done ({len(X_te)} test obs)')

shap_stab = pd.DataFrame(fold_shap_list)
shap_mean = shap_stab.mean().sort_values(ascending=False)
shap_std  = shap_stab.std()

# Plot top 15 with error bars
top15 = shap_mean.head(15)
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(top15.index[::-1], top15.values[::-1],
        xerr=shap_std[top15.index[::-1]].values,
        color='#2E75B6', ecolor='#888888', capsize=4)
ax.set_xlabel('Mean |SHAP| across folds (± 1 std)', fontsize=11)
ax.set_title('M3 Feature Importance Stability\n(expanding-window folds)', fontsize=11, fontweight='bold')
ax.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.savefig(FIG_DIR / 'shap_fold_stability.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved → {FIG_DIR}/shap_fold_stability.png')
```
